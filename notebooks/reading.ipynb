```python
# Separate valid and invalid months
valid = org_FSStats_data[
    org_FSStats_data["_Month"].between(1, 12)
].copy()

invalid = org_FSStats_data[
    ~org_FSStats_data["_Month"].between(1, 12)
].copy()

# Convert the date column
valid["DateForSearch"] = pd.to_datetime(valid["DateForSearch"])
invalid["DateForSearch"] = pd.to_datetime(invalid["DateForSearch"])

# Check whether an invalid record has a matching
# site and date among valid records
check = invalid.merge(
    valid[["_id_site", "DateForSearch"]],
    on=["_id_site", "DateForSearch"],
    how="left",
    indicator=True
)

print(check["_merge"].value_counts())

display(
    check[[
        "_id_site",
        "_Year",
        "_Month",
        "DateForSearch",
        "_merge"
    ]]
)
```

```python
# Compare the three important Q6 variables
activity_cols = [
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

# Match suspicious records with valid records
comparison = invalid.merge(
    valid[
        ["_id_site", "DateForSearch"] + activity_cols
    ],
    on=["_id_site", "DateForSearch"],
    how="left",
    suffixes=("_invalid", "_valid")
)

# Compare activity values
for col in activity_cols:
    comparison[f"{col}_same"] = (
        comparison[f"{col}_invalid"].fillna(-1)
        ==
        comparison[f"{col}_valid"].fillna(-1)
    )

# Check if all three variables match
comparison["all_match"] = comparison[
    [f"{col}_same" for col in activity_cols]
].all(axis=1)

print("Total comparisons:", len(comparison))
print("\nActivity values match:")
print(comparison["all_match"].value_counts())

display(comparison[
    ["_id_site", "DateForSearch", "all_match"]
])
```

```python
# Check which activity variables differ
for col in activity_cols:
    print(f"\n{col}:")
    print(comparison[f"{col}_same"].value_counts())

# Display original and matching values side by side
columns = ["_id_site", "DateForSearch"]

for col in activity_cols:
    columns.extend([
        f"{col}_invalid",
        f"{col}_valid"
    ])

display(comparison[columns])
```

```python
# Known-location 2024 records
known_2024 = q6_2024_clean[
    q6_2024_clean["County"].notna()
].copy()

# Missing-location 2024 records
missing_2024 = q6_2024_clean[
    q6_2024_clean["County"].isna()
].copy()

metrics = ["HouseholdsReg", "IndividualsReg", "PoundsReg"]

results = []

for _, row in missing_2024.iterrows():

    possible = known_2024[
        known_2024["_Month"] == row["_Month"]
    ].copy()

    for col in metrics:
        if pd.isna(row[col]):
            possible = possible[possible[col].isna()]
        else:
            possible = possible[possible[col] == row[col]]

    results.append({
        "_id_site": row["_id_site"],
        "_Month": row["_Month"],
        "HouseholdsReg": row["HouseholdsReg"],
        "IndividualsReg": row["IndividualsReg"],
        "PoundsReg": row["PoundsReg"],
        "exact_matches": len(possible),
        "matched_site": possible["_id_site"].tolist(),
        "matched_county": possible["County"].tolist(),
        "matched_type": possible["AgencyType"].tolist()
    })

duplicate_check = pd.DataFrame(results)

display(duplicate_check)

print(
    "Rows with at least one exact match:",
    (duplicate_check["exact_matches"] > 0).sum()
)

print(
    "Rows with no exact match:",
    (duplicate_check["exact_matches"] == 0).sum()
)
```

```python
# Create indicator columns for broad site categories
site_rows = q6_clean[
    q6_clean["_Month"].between(1, 12)
].copy()

site_rows["is_core_shelf"] = site_rows["AgencyType"].isin([
    "Food Shelf",
    "Brick and Mortar Site",
    "Brick and Mortar Sites"
])

site_rows["is_mobile_outreach"] = site_rows["AgencyType"].isin([
    "Mobile",
    "Outreach Program"
])

site_rows["is_meal_program"] = site_rows["AgencyType"].isin([
    "Meal Program",
    "Meal Programs",
    "On-Site Meal Programs"
])

site_rows["is_college"] = (
    site_rows["AgencyType"] == "College Campus Program"
)

site_rows["is_tribal"] = (
    site_rows["AgencyType"] == "Tribal Program"
)

site_class = (
    site_rows
    .groupby("_id_site")[
        [
            "is_core_shelf",
            "is_mobile_outreach",
            "is_meal_program",
            "is_college",
            "is_tribal"
        ]
    ]
    .max()
    .reset_index()
)

display(site_class.head())

conditions = [
    site_class["is_core_shelf"] & site_class["is_mobile_outreach"],
    site_class["is_core_shelf"],
    site_class["is_mobile_outreach"],
    site_class["is_meal_program"],
    site_class["is_college"],
    site_class["is_tribal"]
]

choices = [
    "Food Shelf / Outreach",
    "Food Shelf",
    "Mobile / Outreach",
    "Meal Program",
    "College Program",
    "Tribal Program"
]

site_class["stable_type"] = np.select(
    conditions,
    choices,
    default="Other"
)

print(site_class["stable_type"].value_counts())
```

```python
# Q6 activity variables
metrics = [
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

# Keep only valid months for now
valid_month_data = q6_clean[
    q6_clean["_Month"].between(1, 12)
].copy()

# Known-location rows
known_all = valid_month_data[
    valid_month_data["County"].notna()
].copy()

# Missing-location rows
missing_all = valid_month_data[
    valid_month_data["County"].isna()
].copy()

# Only use rows where all 3 Q6 measures are reported
known_complete = known_all[
    known_all[metrics].notna().all(axis=1)
].copy()

missing_complete = missing_all[
    missing_all[metrics].notna().all(axis=1)
].copy()

# Give each missing row its original row number
missing_complete = (
    missing_complete
    .reset_index()
    .rename(columns={"index": "original_index"})
)

# Remove duplicate copies on the known side
known_unique = known_complete.drop_duplicates(
    subset=[
        "_id_site",
        "_Year",
        "_Month",
        "HouseholdsReg",
        "IndividualsReg",
        "PoundsReg"
    ]
).copy()

match_columns = [
    "_Year",
    "_Month",
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

matches_all = missing_complete.merge(
    known_unique[
        [
            "_id_site",
            "County"
        ] + match_columns
    ],
    on=match_columns,
    how="left",
    suffixes=("_missing", "_known")
)

# Keep actual matches only
actual_matches = matches_all[
    matches_all["_id_site_known"].notna()
].copy()
match_counts = (
    actual_matches
    .groupby("original_index")["_id_site_known"]
    .nunique()
    .reset_index(name="exact_matches")
)

duplicate_check_all = missing_complete.merge(
    match_counts,
    on="original_index",
    how="left"
)

duplicate_check_all["exact_matches"] = (
    duplicate_check_all["exact_matches"]
    .fillna(0)
    .astype(int)
)

print("Complete missing-location rows:",
      len(duplicate_check_all))

print("\nExactly one match:",
      (duplicate_check_all["exact_matches"] == 1).sum())

print("Multiple matches:",
      (duplicate_check_all["exact_matches"] > 1).sum())

print("No exact match:",
      (duplicate_check_all["exact_matches"] == 0).sum())

year_summary = pd.crosstab(
    duplicate_check_all["_Year"],
    duplicate_check_all["exact_matches"]
)

display(year_summary)
```

```python
metrics = [
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

# Rows with a known county and all 3 Q6 measures observed
known_complete = q6_valid_months[
    q6_valid_months["County"].notna() &
    q6_valid_months[metrics].notna().all(axis=1)
].copy()

# Rows with missing county and all 3 measures observed
missing_complete = q6_valid_months[
    q6_valid_months["County"].isna() &
    q6_valid_months[metrics].notna().all(axis=1)
].copy()

# Preserve original row index so we know which rows to remove
missing_complete = (
    missing_complete
    .reset_index()
    .rename(columns={"index": "source_index"})
)

match_columns = [
    "_Year",
    "_Month",
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

exact_matches = missing_complete.merge(
    known_complete[
        ["_id_site", "County"] + match_columns
    ],
    on=match_columns,
    how="left",
    suffixes=("_missing", "_known")
)

confirmed_duplicates = exact_matches[
    exact_matches["_id_site_known"].notna()
].copy()

duplicate_indices = (
    confirmed_duplicates["source_index"]
    .drop_duplicates()
    .tolist()
)

print("Confirmed blank-county duplicate rows:",
      len(duplicate_indices))

q6_no_duplicates = q6_valid_months[
    ~q6_valid_months.index.isin(duplicate_indices)
].copy()

print("Rows before duplicate removal:",
      len(q6_valid_months))

print("Duplicate rows removed:",
      len(q6_valid_months) - len(q6_no_duplicates))

print("Rows remaining:",
      len(q6_no_duplicates))

remaining_missing = q6_no_duplicates[
    q6_no_duplicates["County"].isna()
].copy()

print("Remaining rows with missing county:",
      len(remaining_missing))

print("Unique unresolved sites:",
      remaining_missing["_id_site"].nunique())

print("\nBy year:")
print(
    remaining_missing["_Year"]
    .value_counts()
    .sort_index()
)
```

```python
unresolved = q6_no_duplicates[
    q6_no_duplicates["County"].isna()
].copy()

print("Agency types:")
print(unresolved["AgencyType"].value_counts(dropna=False))

print("\nFood banks:")
print(unresolved["FoodBank"].value_counts(dropna=False))

for col in ["HouseholdsReg", "IndividualsReg", "PoundsReg"]:

    unresolved_total = unresolved[col].sum()
    overall_total = q6_no_duplicates[col].sum()

    percent = 100 * unresolved_total / overall_total

    print(
        col,
        ":",
        round(unresolved_total, 2),
        "(",
        round(percent, 3),
        "%)"
    )

unresolved_summary = (
    unresolved
    .groupby("_id_site")
    .agg(
        rows=("_Month", "count"),
        first_year=("_Year", "min"),
        last_year=("_Year", "max"),
        households=("HouseholdsReg", "sum"),
        individuals=("IndividualsReg", "sum"),
        pounds=("PoundsReg", "sum")
    )
    .reset_index()
)

display(unresolved_summary)
```

```python
# Start from the data after:
# 1. removing invalid-month duplicates
# 2. removing confirmed blank-county duplicate feeds

site_month_clean = q6_no_duplicates.copy()

# Add the stable site classification we created earlier
site_month_clean = site_month_clean.merge(
    site_class[
        ["_id_site", "stable_type"]
    ],
    on="_id_site",
    how="left"
)

# Add official Organization flags
org_flags = org_data[
    [
        "__ID",
        "isDistributionSite",
        "isAgency",
        "isActive",
        "block_geoid"
    ]
].copy()

site_month_clean = site_month_clean.merge(
    org_flags,
    left_on="_id_site",
    right_on="__ID",
    how="left",
    suffixes=("", "_org")
)

# Flag whether geographic location is usable
site_month_clean["location_resolved"] = (
    site_month_clean["County"].notna()
)
print("Final cleaned rows:", len(site_month_clean))

print("\nYears:")
print(
    site_month_clean["_Year"]
    .value_counts()
    .sort_index()
)

print("\nLocation resolved:")
print(
    site_month_clean["location_resolved"]
    .value_counts()
)

print("\nStable site types:")
print(
    site_month_clean["stable_type"]
    .value_counts(dropna=False)
)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
org_FSStats_data = pd.read_excel("../data/org_FSStats.xlsx")

print("Rows and columns:", org_FSStats_data.shape)
print("\nColumn names:")
print(org_FSStats_data.columns.tolist())

org_FSStats_data.head()
```

```python
q6_clean = org_FSStats_data.copy()

# Standardize county names
county_mapping = {
    "Hennepin-Minneapolis": "Hennepin",
    "Hennepin-Suburbs": "Hennepin",
    "Ramsey-Saint Paul": "Ramsey",
    "Ramsey-Suburbs": "Ramsey",
    "Saint Louis": "St. Louis"
}

q6_clean["County"] = (
    q6_clean["County"]
    .str.strip()
    .replace(county_mapping)
)
```

```python
org_data = pd.read_excel("../data/Organization.xlsx")

print("Shape:", org_data.shape)
print("\nColumns:")
print(org_data.columns.tolist())

org_data.head()
```

```python
# Create indicator columns for broad site categories
site_rows = q6_clean[
    q6_clean["_Month"].between(1, 12)
].copy()

site_rows["is_core_shelf"] = site_rows["AgencyType"].isin([
    "Food Shelf",
    "Brick and Mortar Site",
    "Brick and Mortar Sites"
])

site_rows["is_mobile_outreach"] = site_rows["AgencyType"].isin([
    "Mobile",
    "Outreach Program"
])

site_rows["is_meal_program"] = site_rows["AgencyType"].isin([
    "Meal Program",
    "Meal Programs",
    "On-Site Meal Programs"
])

site_rows["is_college"] = (
    site_rows["AgencyType"] == "College Campus Program"
)

site_rows["is_tribal"] = (
    site_rows["AgencyType"] == "Tribal Program"
)

site_class = (
    site_rows
    .groupby("_id_site")[
        [
            "is_core_shelf",
            "is_mobile_outreach",
            "is_meal_program",
            "is_college",
            "is_tribal"
        ]
    ]
    .max()
    .reset_index()
)

display(site_class.head())

conditions = [
    site_class["is_core_shelf"] & site_class["is_mobile_outreach"],
    site_class["is_core_shelf"],
    site_class["is_mobile_outreach"],
    site_class["is_meal_program"],
    site_class["is_college"],
    site_class["is_tribal"]
]

choices = [
    "Food Shelf / Outreach",
    "Food Shelf",
    "Mobile / Outreach",
    "Meal Program",
    "College Program",
    "Tribal Program"
]

site_class["stable_type"] = np.select(
    conditions,
    choices,
    default="Other"
)

print(site_class["stable_type"].value_counts())
```

```python
# Invalid month records
metrics = [
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]
invalid_month_rows = q6_clean[
    ~q6_clean["_Month"].between(1, 12)
].copy()

month0 = invalid_month_rows[
    invalid_month_rows["_Month"] == 0
].copy()

month15 = invalid_month_rows[
    invalid_month_rows["_Month"] == 15
].copy()

print("Month 0 rows:", len(month0))
print("Month 15 rows:", len(month15))

dec_2023 = q6_clean[
    (q6_clean["_Year"] == 2023) &
    (q6_clean["_Month"] == 12)
][
    [
        "_id_site",
        "HouseholdsReg",
        "IndividualsReg",
        "PoundsReg"
    ]
].copy()

month0_check = month0.merge(
    dec_2023,
    on="_id_site",
    how="left",
    suffixes=("_month0", "_dec2023")
)

for col in metrics:
    month0_check[col + "_same"] = (
        month0_check[col + "_month0"].fillna(-999999)
        ==
        month0_check[col + "_dec2023"].fillna(-999999)
    )

same_columns = [
    "HouseholdsReg_same",
    "IndividualsReg_same",
    "PoundsReg_same"
]

month0_check["all_three_same"] = (
    month0_check[same_columns].all(axis=1)
)

print("Month-0 rows matching December 2023 exactly:")
print(month0_check["all_three_same"].value_counts())

display(
    month0_check[
        [
            "_id_site",
            "HouseholdsReg_month0",
            "HouseholdsReg_dec2023",
            "IndividualsReg_month0",
            "IndividualsReg_dec2023",
            "PoundsReg_month0",
            "PoundsReg_dec2023",
            "all_three_same"
        ]
    ]
)

march_2024 = q6_clean[
    (q6_clean["_Year"] == 2024) &
    (q6_clean["_Month"] == 3)
][
    [
        "_id_site",
        "HouseholdsReg",
        "IndividualsReg",
        "PoundsReg"
    ]
].copy()

month15_check = month15.merge(
    march_2024,
    on="_id_site",
    how="left",
    suffixes=("_month15", "_mar2024")
)

display(month15_check)
```

```python
# Remove the 26 invalid-month duplicate records
valid_month_mask = q6_clean["_Month"].between(1, 12)

q6_valid_months = q6_clean[
    valid_month_mask
].copy()

print("Original rows:", len(q6_clean))
print("Rows removed:", len(q6_clean) - len(q6_valid_months))
print("Rows remaining:", len(q6_valid_months))

print(
    "Invalid months remaining:",
    (~q6_valid_months["_Month"].between(1, 12)).sum()
)
```

```python
metrics = [
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

# Rows with a known county and all 3 Q6 measures observed
known_complete = q6_valid_months[
    q6_valid_months["County"].notna() &
    q6_valid_months[metrics].notna().all(axis=1)
].copy()

# Rows with missing county and all 3 measures observed
missing_complete = q6_valid_months[
    q6_valid_months["County"].isna() &
    q6_valid_months[metrics].notna().all(axis=1)
].copy()

# Preserve original row index so we know which rows to remove
missing_complete = (
    missing_complete
    .reset_index()
    .rename(columns={"index": "source_index"})
)

match_columns = [
    "_Year",
    "_Month",
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

exact_matches = missing_complete.merge(
    known_complete[
        ["_id_site", "County"] + match_columns
    ],
    on=match_columns,
    how="left",
    suffixes=("_missing", "_known")
)

confirmed_duplicates = exact_matches[
    exact_matches["_id_site_known"].notna()
].copy()

duplicate_indices = (
    confirmed_duplicates["source_index"]
    .drop_duplicates()
    .tolist()
)

print("Confirmed blank-county duplicate rows:",
      len(duplicate_indices))

q6_no_duplicates = q6_valid_months[
    ~q6_valid_months.index.isin(duplicate_indices)
].copy()

print("Rows before duplicate removal:",
      len(q6_valid_months))

print("Duplicate rows removed:",
      len(q6_valid_months) - len(q6_no_duplicates))

print("Rows remaining:",
      len(q6_no_duplicates))

remaining_missing = q6_no_duplicates[
    q6_no_duplicates["County"].isna()
].copy()

print("Remaining rows with missing county:",
      len(remaining_missing))

print("Unique unresolved sites:",
      remaining_missing["_id_site"].nunique())

print("\nBy year:")
print(
    remaining_missing["_Year"]
    .value_counts()
    .sort_index()
)
```

```python
unresolved = q6_no_duplicates[
    q6_no_duplicates["County"].isna()
].copy()

print("Agency types:")
print(unresolved["AgencyType"].value_counts(dropna=False))

print("\nFood banks:")
print(unresolved["FoodBank"].value_counts(dropna=False))

for col in ["HouseholdsReg", "IndividualsReg", "PoundsReg"]:

    unresolved_total = unresolved[col].sum()
    overall_total = q6_no_duplicates[col].sum()

    percent = 100 * unresolved_total / overall_total

    print(
        col,
        ":",
        round(unresolved_total, 2),
        "(",
        round(percent, 3),
        "%)"
    )

unresolved_summary = (
    unresolved
    .groupby("_id_site")
    .agg(
        rows=("_Month", "count"),
        first_year=("_Year", "min"),
        last_year=("_Year", "max"),
        households=("HouseholdsReg", "sum"),
        individuals=("IndividualsReg", "sum"),
        pounds=("PoundsReg", "sum")
    )
    .reset_index()
)

display(unresolved_summary)
```

```python
# Start from the data after:
# 1. removing invalid-month duplicates
# 2. removing confirmed blank-county duplicate feeds

site_month_clean = q6_no_duplicates.copy()

# Add the stable site classification we created earlier
site_month_clean = site_month_clean.merge(
    site_class[
        ["_id_site", "stable_type"]
    ],
    on="_id_site",
    how="left"
)

# Add official Organization flags
org_flags = org_data[
    [
        "__ID",
        "isDistributionSite",
        "isAgency",
        "isActive",
        "block_geoid"
    ]
].copy()

site_month_clean = site_month_clean.merge(
    org_flags,
    left_on="_id_site",
    right_on="__ID",
    how="left",
    suffixes=("", "_org")
)

# Flag whether geographic location is usable
site_month_clean["location_resolved"] = (
    site_month_clean["County"].notna()
)
print("Final cleaned rows:", len(site_month_clean))

print("\nYears:")
print(
    site_month_clean["_Year"]
    .value_counts()
    .sort_index()
)

print("\nLocation resolved:")
print(
    site_month_clean["location_resolved"]
    .value_counts()
)

print("\nStable site types:")
print(
    site_month_clean["stable_type"]
    .value_counts(dropna=False)
)
```

```python
sites_to_check = [1189, 1365]

kanabec_check = site_month_clean[
    site_month_clean["_id_site"].isin(sites_to_check)
].copy()

counts = (
    kanabec_check
    .groupby(["_id_site", "_Year", "_Month"])
    .size()
    .reset_index(name="n_rows")
)

repeated_months = counts[
    counts["n_rows"] > 1
][
    ["_id_site", "_Year", "_Month"]
]

kanabec_duplicates = kanabec_check.merge(
    repeated_months,
    on=["_id_site", "_Year", "_Month"],
    how="inner"
)

display(
    kanabec_duplicates[
        [
            "_id_site",
            "_Year",
            "_Month",
            "__ID",
            "HouseholdsReg",
            "IndividualsReg",
            "PoundsReg"
        ]
    ].sort_values(
        ["_id_site", "_Year", "_Month", "__ID"]
    )
)

duplicate_value_check = (
    kanabec_duplicates
    .groupby(["_id_site", "_Year", "_Month"])
    .agg(
        household_values=("HouseholdsReg", "nunique"),
        individual_values=("IndividualsReg", "nunique"),
        pound_values=("PoundsReg", "nunique"),
        row_count=("__ID", "count")
    )
    .reset_index()
)

display(duplicate_value_check)

site_month_clean.to_csv(
    "../data/processed/site_month_clean.csv",
    index=False
)
```

The original activity file contained 28,287 rows. Twenty-six invalid-month records and 87 blank-county duplicate-feed records were removed. One exact duplicate site-month record for site 1189 was removed, followed by 19 rows belonging to six persistent mirror-site IDs. Forty-five rows across 16 sites still lack reliable county information; these are retained in the master file but excluded from county-level analyses. Seven implausible household-count entries were corrected using each site's median household count calculated from non-flagged rows. The final cleaned dataset contains 28,154 rows

```python
metrics = [
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

match_columns = [
    "_Year",
    "_Month",
    "HouseholdsReg",
    "IndividualsReg",
    "PoundsReg"
]

# Use valid-month data before blank-county duplicates
# were removed, so the mirror evidence is still present.
mirror_base = q6_valid_months[
    q6_valid_months[metrics].notna().all(axis=1) &
    (q6_valid_months["HouseholdsReg"] >= 20) &
    (q6_valid_months["PoundsReg"] > 0)
][
    ["_id_site"] + match_columns
].copy()

mirror_base = mirror_base.drop_duplicates()
```

```python
site_pairs = mirror_base.merge(
    mirror_base,
    on=match_columns,
    how="inner",
    suffixes=("_a", "_b")
)

# Remove self-matches and reversed duplicates
site_pairs = site_pairs[
    site_pairs["_id_site_a"] < site_pairs["_id_site_b"]
].copy()

site_pairs["year_month"] = (
    site_pairs["_Year"].astype(str)
    + "-"
    + site_pairs["_Month"].astype(str).str.zfill(2)
)

pair_summary = (
    site_pairs
    .groupby(
        ["_id_site_a", "_id_site_b"],
        as_index=False
    )
    .agg(
        matching_months=("year_month", "nunique")
    )
)

# Require at least 3 matching months
mirror_pairs = pair_summary[
    pair_summary["matching_months"] >= 3
].copy()
```

```python
mirror_ids = [
    15361,
    15362,
    15363,
    15364,
    15365,
    15366
]

mirror_evidence = mirror_pairs[
    mirror_pairs["_id_site_a"].isin(mirror_ids) |
    mirror_pairs["_id_site_b"].isin(mirror_ids)
].sort_values(
    "matching_months",
    ascending=False
)

display(mirror_evidence)
```

```python
mirror_rows = site_month_clean[
    site_month_clean["_id_site"].isin(mirror_ids)
].copy()

print("Mirror rows removed:", len(mirror_rows))

site_month_clean = site_month_clean[
    ~site_month_clean["_id_site"].isin(mirror_ids)
].copy()

print("Rows after mirror removal:", len(site_month_clean))
```

```python
# Detect extreme HouseholdReg values using three independent checks.

site_household_median = (
    site_month_clean
    .groupby("_id_site")[
        "HouseholdsReg"
    ]
    .transform("median")
)

rule_households_vs_individuals = (
    site_month_clean["HouseholdsReg"]
    >
    1.5
    * site_month_clean["IndividualsReg"]
)

rule_households_equal_pounds = (
    site_month_clean["HouseholdsReg"].notna()
    &
    site_month_clean["PoundsReg"].notna()
    &
    (
        site_month_clean["HouseholdsReg"]
        ==
        site_month_clean["PoundsReg"]
    )
)

rule_extreme_vs_site = (
    (
        site_month_clean["HouseholdsReg"]
        >
        10 * site_household_median
    )
    &
    (
        site_month_clean["HouseholdsReg"]
        > 500
    )
)

outlier_rule_count = (
    rule_households_vs_individuals.astype(int)
    +
    rule_households_equal_pounds.astype(int)
    +
    rule_extreme_vs_site.astype(int)
)

household_outlier = (
    outlier_rule_count >= 2
)

outlier_audit = site_month_clean.loc[
    household_outlier,
    [
        "_id_site",
        "_Year",
        "_Month",
        "County",
        "HouseholdsReg",
        "IndividualsReg",
        "PoundsReg"
    ]
].copy()

outlier_audit[
    "rules_triggered"
] = outlier_rule_count[
    household_outlier
].to_numpy()

print(
    "Household outliers detected:",
    len(outlier_audit)
)

display(
    outlier_audit.sort_values(
        [
            "_Year",
            "_Month",
            "_id_site"
        ]
    )
)
```

```python
# Build site medians using only non-outlier rows
household_reference = site_month_clean[
    ~household_outlier
][
    [
        "_id_site",
        "HouseholdsReg"
    ]
].copy()

site_household_medians = (
    household_reference
    .groupby("_id_site")[
        "HouseholdsReg"
    ]
    .median()
)

# Create an audit table before changing anything
household_outlier_audit = site_month_clean.loc[
    household_outlier,
    [
        "_id_site",
        "_Year",
        "_Month",
        "County",
        "HouseholdsReg",
        "IndividualsReg",
        "PoundsReg"
    ]
].copy()

household_outlier_audit = (
    household_outlier_audit
    .rename(
        columns={
            "HouseholdsReg":
                "original_households"
        }
    )
)

household_outlier_audit[
    "replacement_households"
] = (
    household_outlier_audit[
        "_id_site"
    ]
    .map(
        site_household_medians
    )
)

print(
    "Outliers to replace:",
    len(
        household_outlier_audit
    )
)

print(
    "Missing replacement medians:",
    household_outlier_audit[
        "replacement_households"
    ]
    .isna()
    .sum()
)

display(
    household_outlier_audit[
        [
            "_id_site",
            "_Year",
            "_Month",
            "County",
            "original_households",
            "replacement_households",
            "IndividualsReg",
            "PoundsReg"
        ]
    ]
)
```

```python
replacement_values = (
    site_month_clean.loc[
        household_outlier,
        "_id_site"
    ]
    .map(
        site_household_medians
    )
)

site_month_clean.loc[
    household_outlier,
    "HouseholdsReg"
] = (
    replacement_values.to_numpy()
)

display(
    site_month_clean.loc[
        household_outlier,
        [
            "_id_site",
            "_Year",
            "_Month",
            "County",
            "HouseholdsReg",
            "IndividualsReg",
            "PoundsReg"
        ]
    ]
)
```

```python
site_household_median_check = (
    site_month_clean
    .groupby("_id_site")[
        "HouseholdsReg"
    ]
    .transform("median")
)

check_rule_1 = (
    site_month_clean[
        "HouseholdsReg"
    ]
    >
    1.5
    * site_month_clean[
        "IndividualsReg"
    ]
)

check_rule_2 = (
    site_month_clean[
        "HouseholdsReg"
    ].notna()
    &
    site_month_clean[
        "PoundsReg"
    ].notna()
    &
    (
        site_month_clean[
            "HouseholdsReg"
        ]
        ==
        site_month_clean[
            "PoundsReg"
        ]
    )
)

check_rule_3 = (
    (
        site_month_clean[
            "HouseholdsReg"
        ]
        >
        10
        * site_household_median_check
    )
    &
    (
        site_month_clean[
            "HouseholdsReg"
        ]
        > 500
    )
)

remaining_bad_rows = (
    check_rule_1.astype(int)
    +
    check_rule_2.astype(int)
    +
    check_rule_3.astype(int)
) >= 2

print(
    "Remaining household outliers:",
    remaining_bad_rows.sum()
)
```
